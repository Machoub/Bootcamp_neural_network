```python
import pandas as pd
import seaborn as sns

df = pd.read_csv('Salary_dataset.csv')
print(df)
```

```python
sns.set_theme(style="darkgrid")
sns.jointplot(
    data=df,
    x="YearsExperience",
    y="Salary",
    kind="scatter", color="m",
    marginal_kws=dict(fill=True, color="m", alpha=0.5)
)
```

# Task
Load the CSV file "Salary_Data.csv" into a pandas DataFrame, create a scatter plot using seaborn with 'YearsExperience' on the x-axis and 'Salary' on the y-axis, customize the plot with purple lines, light purple background, and purple points, train a linear regression model on the data, and display the regression line on the scatter plot.

## Prepare data for modeling

### Subtask:
Separate the features (YearsExperience) and the target (Salary) into arrays.


**Reasoning**:
Separate the features (YearsExperience) and the target (Salary) into arrays.



```python
X = df['YearsExperience'].values.reshape(-1, 1)
y = df['Salary'].values
```

## Train a linear regression model

### Subtask:
Use scikit-learn to train a linear regression model on the data.


**Reasoning**:
Train a linear regression model using the prepared data.



```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)
model.score(X, y)
predicted_salary = model.predict(X)
print(predicted_salary)
```

## Visualize the model

### Subtask:
Modify the existing plot to include the linear regression line from the trained model.


**Reasoning**:
Generate predictions, create a scatter plot with the regression line, and customize the plot appearance as per the instructions.



```python
sns.regplot(
    data=df,
    x="YearsExperience",
    y="Salary",
    ci=95,               # intervalle de confiance
    scatter_kws={"alpha":0.7},  # style des points
    line_kws={"color":"blue"}   # style de la droite
).set(
    xlabel="Years of Experience",
    ylabel="Salary",
    title="Linear Regression: Years of Experience vs Salary"
)
```

```python
ten_years_experience = 10
predicted_salary = model.predict([[ten_years_experience]])
print(f"Predicted salary for {ten_years_experience} years of experience: ${predicted_salary[0]}")
```
